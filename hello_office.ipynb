```python
names = input("Enter the names of everyone in the office, separated by commas: ").split(",")

for name in names:
    name = name.strip()
    if name:
        print(f"Hello, {name}!")
```

```python
# add today's date to the greeting
from datetime import datetime

today = datetime.now().date()

names = input("Enter the names of everyone in the office, separated by commas: ").split(",")

for name in names:
    name = name.strip()
    if name:
        print(f"Good luck, {name}! Today is {today}.")
```

```python
# save greetings to a text file
with open("greetings.txt", "w") as f:
    for name in names:
        name = name.strip()
        if name:
            f.write(f"Good luck, {name}! Today is {today}.\n")
```
